```python
import os
import tarfile
import lzma
import gzip
import shutil
import zstandard as zstd

def compress_folder_lzma(folder_path, output_file):
    """Compress the entire folder using LZMA (.tar.xz) - Best Compression"""
    with tarfile.open(output_file, "w:xz") as tar:
        tar.add(folder_path, arcname=os.path.basename(folder_path))

def compress_folder_gzip(folder_path, output_file):
    """Compress the entire folder using Gzip (.tar.gz) - Faster"""
    with tarfile.open(output_file, "w:gz") as tar:
        tar.add(folder_path, arcname=os.path.basename(folder_path))

def compress_folder_zstd(folder_path, output_file):
    """Compress the entire folder using Zstandard (.tar.zst) - Fastest"""
    with tarfile.open("temp.tar", "w") as tar:
        tar.add(folder_path, arcname=os.path.basename(folder_path))
    
    # Now apply Zstd compression
    compressor = zstd.ZstdCompressor(level=22)  # Max compression level
    with open("temp.tar", "rb") as f_in, open(output_file, "wb") as f_out:
        f_out.write(compressor.compress(f_in.read()))
    
    os.remove("temp.tar")  # Cleanup temporary tar file

# Example usage
model_folder = "/home/muhammad-osama/Downloads/modern_bert"  # Change this to your actual folder name

compress_folder_lzma(model_folder, "/home/muhammad-osama/Downloads/modern_bert.tar.xz")
# compress_folder_gzip(model_folder, "models_compressed.tar.gz")
# compress_folder_zstd(model_folder, "models_compressed.tar.zst")

print("Compression completed! ✅")
```

## Uncompress

```python
import tarfile

def extract_lzma(archive_path, output_folder):
    """Extract a .tar.xz compressed archive."""
    with tarfile.open(archive_path, "r:xz") as tar:
        tar.extractall(path=output_folder)
    print(f"Extracted {archive_path} to {output_folder}")

# Example usage
extract_lzma("models_compressed.tar.xz", "extracted_models")
```
